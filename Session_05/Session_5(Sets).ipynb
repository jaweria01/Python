## **Sets**

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

***Characterstics:***

Unordered

Mutable

No Duplicates

Can't contain mutable data types

```python
# unordered means that 1,2,3 = 3,2,1
# Sets itself are mutable , but they can not take mutable data types like sets inside sets will not work / 2D sets, or list inside sets
```

## **Set Operation**

```python
# Union(|)
# print unique items of both sets
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1 | s2
```

```python
# Intersation
# print common items of both sets
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1 & s2
```

```python
# Difference(-)
# print items of s1 which are not present in s2
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1 - s2
```

```python
# Symmetric Difference(^)
# print all except common, in union all items will be print and which are common will be print once thats the difference
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1 ^ s2
```

```python
# Above 4 operations are basics
```

```python
# Membership Test
s1 = {1,2,3,4,5}
1 not in s1
```

```python
# Iteration/loops
s1 = {10,20,30,40,50}
for i in s1:
  print(i)
```

## **Sets Functions**

```python
# len/sum/min/max/sorted
# union()
# intersaction()
# difference()
# symmetric_difference()
# update()...>union update
# intersation_update()
# difference_update()
# symmetric_difference_update()
# isdisjoint()
#issubset()
# issuperset()
```

```python
# len/sum/min/max/sorted
s = {3,1,4,5,2,7}
len(s)
sum(s)
min(s)
max(s)
sorted(s,reverse=True)
```

```python
# union func
a1 = {1,2,3,4,5}
a2 = {4,5,6,7,8}
a1.union(a2)
```

```python
#union update func.
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1.update(s2)
print(s1)
print(s2)
# firstlt, union operation will run inside the update func, which is invisible
# firstly , we print s1, the result of union operation will put in s1, and s2 will remain same as it is., a permanent change will be in s1
# then we print s2 that was earlier
```

```python
# intersection
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.intersection(s2)
```

```python
#intersection_update

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
# the result of intersaction will store in s1 and s2 will be remain same
s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# difference
## print items of s1 which are not present in s2
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.difference(s2)
```

```python
#/difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1.difference_update(s2)
print(s1)
print(s2)
```

```python
# symmetric_difference
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2)
```

```python
# symmetric_difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
# s1 contain all except common in both
s1.symmetric_difference_update(s2)
print(s2)
print(s1)
```

```python
# isdisjoint/issubset/issuperset
# disjoint sets are those which have no common items
s1 = {1,2,3,4}
s2 = {7,8,5,6}

s1.isdisjoint(s2)
```

```python
# issubset()
# subset()
# a set inside another set. s1 inside s2 set
s1 = {1,2,3,4,5}
s2 = {3,4,5}
# s2 is subset of s2
s2.issubset(s1)
```

```python
# issuperser()
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s1.issuperset(s2)
```

```python
# Both have different memory address
# copy
s1 = {1,2,3}
s2 = s1.copy()

print(s1)
print(s2)
```

## **Frozenset**

Frozen set is just an immutable version of a Python set object

No item will be added and no can be deleted

```python
# create frozenset
# inside frozen set , can be any iterable like tuple , list, set
fs1 = frozenset([1,2,3])
print(fs1)
```

```python
fs1 = frozenset([1,2,3])
fs2 = frozenset([3,4,5])

fs1 | fs2
```

```python
# what works and what does not
# works -> all read functions
# does't work -> write operations like add and update function will not work which work in sets()
```

```python
# When to use >> when you make an application where no changes are required. then you will use frozensets() otherwise you use sets
# 2D sets
fs = frozenset([1,2,frozenset([3,4])])
fs
```

```python
fs = frozenset([1,2,frozenset((3,4))])
fs
```
